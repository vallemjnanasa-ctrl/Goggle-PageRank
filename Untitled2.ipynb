```python
import numpy as np
from IPython.display import display, HTML

print("Google PageRank Mini")
print("Libraries loaded successfully!")
```

```python
# Five web pages
pages = ["A", "B", "C", "D", "E"]

# Links between the pages
links = {
    "A": ["B", "C"],
    "B": ["C", "D"],
    "C": ["A", "E"],
    "D": ["C", "E"],
    "E": ["A", "D"]
}

print("WEB PAGES AND THEIR LINKS")
print("-" * 35)

for page in pages:
    print(f"Page {page} → {', '.join(links[page])}")
```

```python
# Number of pages
n = len(pages)

# Create an empty matrix
M = np.zeros((n, n))

# Fill transition probabilities
for j, page in enumerate(pages):
    destinations = links[page]

    for destination in destinations:
        i = pages.index(destination)
        M[i, j] = 1 / len(destinations)

print("TRANSITION PROBABILITY MATRIX")
print()

print("       A      B      C      D      E")

for i, page in enumerate(pages):
    print(
        f"{page}   "
        + "  ".join(f"{value:.2f}" for value in M[i])
    )
```

```python
# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(M)

print("EIGENVALUES")
print("-" * 30)

for value in eigenvalues:
    print(round(np.real(value), 6))

# Find the eigenvalue closest to 1
index = np.argmin(np.abs(eigenvalues - 1))

eigenvalue = np.real(eigenvalues[index])

# Corresponding eigenvector
pagerank = np.real(eigenvectors[:, index])

# Make values positive
pagerank = np.abs(pagerank)

# Normalize so that sum = 1
pagerank = pagerank / np.sum(pagerank)

print("\nEigenvalue closest to 1:")
print(round(eigenvalue, 6))

print("\nCorresponding Eigenvector / PageRank:")
for page, score in zip(pages, pagerank):
    print(f"Page {page} = {score:.6f}")
```

```python
# Initial PageRank
rank = np.ones(n) / n

# Number of iterations
iterations = 20

print("POWER METHOD")
print("-" * 40)

for i in range(iterations):
    rank = M @ rank
    rank = rank / np.sum(rank)

    print(
        f"Iteration {i+1:2d}: ",
        np.round(rank, 6)
    )

print("\nFinal PageRank using Power Method:")

for page, score in zip(pages, rank):
    print(f"Page {page} = {score:.6f}")
```

```python
# Sort pages from highest PageRank to lowest
ranking = sorted(
    zip(pages, pagerank),
    key=lambda x: x[1],
    reverse=True
)

print("FINAL PAGE RANKING")
print("=" * 40)

for position, (page, score) in enumerate(ranking, 1):
    print(
        f"{position}. Page {page} "
        f"→ PageRank = {score:.4f}"
    )
```
